### Import thư viện và đọc dữ liệu

```python
import time
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
%matplotlib inline

df = pd.read_excel("../Dataset/Online Retail.xlsx")
```

### Các bước xử lý của câu 1 → 3 (chuẩn bị dữ liệu)

```python
df["Description"] = df["Description"].str.strip()
df = df.dropna(axis=0, subset=["InvoiceNo"])
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

df = df[~df["InvoiceNo"].str.contains("C")]
```

### Câu 4: Thống kê số dòng dữ liệu theo từng quốc gia

```python
country_counts = df["Country"].value_counts()
country_counts
```

```python
country_counts.plot(kind="bar", figsize=(14, 5), logy=True)
plt.title("Số dòng dữ liệu theo quốc gia (thang log)")
plt.ylabel("Số dòng")
plt.show()
```
