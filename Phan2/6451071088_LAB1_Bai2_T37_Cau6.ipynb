### Import thư viện và đọc dữ liệu

```python
import time
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
from mlxtend.frequent_patterns import apriori, fpgrowth, association_rules
%matplotlib inline

df = pd.read_excel("../Dataset/Online Retail.xlsx")
```

### Các bước xử lý của câu 1 → 5 (chuẩn bị dữ liệu)

```python
df["Description"] = df["Description"].str.strip()
df = df.dropna(axis=0, subset=["InvoiceNo"])
df["InvoiceNo"] = df["InvoiceNo"].astype(str)

df = df[~df["InvoiceNo"].str.contains("C")]

grouped = (df[df["Country"] == "United Kingdom"]
           .groupby(["InvoiceNo", "Description"])["Quantity"]
           .sum())
```

### Câu 6: Chuyển dữ liệu về dạng hot encoding, mỗi dòng là một hóa đơn

```python
basket = grouped.unstack().reset_index().fillna(0).set_index("InvoiceNo")
```
