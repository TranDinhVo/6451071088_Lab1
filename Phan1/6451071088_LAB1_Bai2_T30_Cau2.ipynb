### Import thư viện và đọc dữ liệu

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.read_excel("../Dataset/tb_tracking.xlsx", header=None)
```

### Các bước xử lý của câu 1 → 1 (chuẩn bị dữ liệu)

```python
df.columns = ["id", "device_id", "co_level", "humidity", "temperature", "time"]
df["time"] = pd.to_datetime(df["time"])
```

### Câu 2: b) Chỉ giữ lại dữ liệu từ ngày 03/01/2018

```python
print("Trước khi lọc:", df.shape)
```

```python
df = df[df["time"] >= "2018-01-03"].reset_index(drop=True)
```

```python
print("Sau khi lọc:", df.shape)
print("Thời gian nhỏ nhất:", df["time"].min())
df.head()
```
