### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/bank/bank.csv", sep=";")
```

### Các bước xử lý của câu 1 → 5 (chuẩn bị dữ liệu)

```python
# Trong bộ dữ liệu, giá trị thiếu được ghi là 'unknown' -> chuyển về NaN
df = df.replace("unknown", np.nan)

df["subscribed"] = (df["y"] == "yes").astype(int)

df["contacts"] = df["campaign"] + df["previous"]          # tổng số lần liên hệ
df["contacted_before"] = (df["previous"] > 0).astype(int)  # đã từng được liên hệ ở chiến dịch trước
```

### Câu 6: Tiền xử lý các thuộc tính thiếu dữ liệu và xóa bỏ các thuộc tính không cần thiết

```python
# poutcome thiếu ~82% và pdays = -1 (chưa liên hệ) chiếm đa số
print((df["pdays"] == -1).mean().round(3))
df["poutcome"].value_counts(dropna=False)
```

```python
df = df.drop(columns=["poutcome", "pdays", "day"])
```

```python
print(df["job"].describe())
print(df["education"].describe())
```

```python
# job, education: điền bằng mode; contact: xem giá trị thiếu là một loại mới 'other'
df["job"] = df["job"].fillna(df["job"].mode()[0])
df["education"] = df["education"].fillna(df["education"].mode()[0])
df["contact"] = df["contact"].fillna("other")
```

```python
print(df.isnull().sum())
df.head()
```

### Nhận xét
- Xóa poutcome (thiếu 82%), pdays (đa số là -1) và day (ít ý nghĩa).
- job, education điền bằng mode; contact thiếu được xem là một loại mới 'other' → không còn giá trị thiếu.
