### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/bank/bank.csv", sep=";")
```

### Câu 2: Liệt kê 5 thuộc tính có số lượng giá trị rỗng nhiều nhất (giá trị 'unknown' được xem là rỗng)

```python
# Trong bộ dữ liệu, giá trị thiếu được ghi là 'unknown' -> chuyển về NaN
df = df.replace("unknown", np.nan)
```

```python
total = df.isnull().sum().sort_values(ascending=False)
percent = (df.isnull().sum() / df.isnull().count() * 100).sort_values(ascending=False)
missing_data = pd.concat([total, percent.round(1)], axis=1, keys=["Total", "%"])
missing_data.head(5)
```

### Nhận xét
- Xem 'unknown' là giá trị rỗng, 5 thuộc tính thiếu nhiều nhất: **poutcome (82%)**, contact (29.3%), education (4.1%), job (0.8%), các thuộc tính còn lại không thiếu.
- poutcome thiếu quá nhiều nên không nên dùng trực tiếp.
