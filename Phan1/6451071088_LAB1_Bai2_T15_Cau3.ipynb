### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/titanic.csv")
```

### Câu 3: Khảo sát sự liên quan giữa Độ tuổi (age), Giới tính (sex) và khả năng sống sót (survived)

```python
survived = "survived"
not_survived = "not survived"
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))
women = df[df["sex"] == "female"]
men = df[df["sex"] == "male"]

ax = sns.histplot(women[women["survived"] == 1]["age"].dropna(), bins=18, label=survived, ax=axes[0], color="tab:green", alpha=0.6)
ax = sns.histplot(women[women["survived"] == 0]["age"].dropna(), bins=40, label=not_survived, ax=axes[0], color="tab:red", alpha=0.6)
ax.legend()
ax.set_title("Female")

ax = sns.histplot(men[men["survived"] == 1]["age"].dropna(), bins=18, label=survived, ax=axes[1], color="tab:green", alpha=0.6)
ax = sns.histplot(men[men["survived"] == 0]["age"].dropna(), bins=40, label=not_survived, ax=axes[1], color="tab:red", alpha=0.6)
ax.legend()
ax.set_title("Male")
plt.show()
```

```python
# Tỷ lệ sống sót theo nhóm tuổi và giới tính (hỗ trợ nhận xét)
age_bins = pd.cut(df["age"], bins=[0, 10, 18, 27, 40, 60, 80])
df.pivot_table(index=age_bins, columns="sex", values="survived", aggfunc="mean", observed=False).round(3)
```

### Nhận xét
- Nam từ 0-10 tuổi có xác suất sống sót cao (~55.6%), sau 10 tuổi tỷ lệ của nam giảm mạnh (chỉ ~12-19%).
- Nữ có tỷ lệ sống sót cao ở mọi lứa tuổi (61-86%), đặc biệt từ 12 tuổi trở lên.
- Nam có số lượng không sống sót nhiều nhất ở độ tuổi khoảng 18-30.
- Nhìn chung tỷ lệ sống sót của nữ cao hơn hẳn nam ở mọi lứa tuổi → tuổi và giới tính đều liên quan đến khả năng sống sót.
