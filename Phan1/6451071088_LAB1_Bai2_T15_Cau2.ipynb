### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/titanic.csv")
```

### Câu 2: Liệt kê 5 thuộc tính có số lượng giá trị rỗng nhiều nhất

```python
total = df.isnull().sum().sort_values(ascending=False)
percent = (df.isnull().sum() / df.isnull().count() * 100).sort_values(ascending=False)
missing_data = pd.concat([total, percent.round(1)], axis=1, keys=["Total", "%"])
missing_data.head(5)
```

### Nhận xét
- 5 thuộc tính thiếu nhiều nhất: **body (90.8%)**, cabin (77.5%), boat (62.9%), home.dest (43.1%), age (20.1%).
- body thiếu rất nghiêm trọng nên sẽ ảnh hưởng lớn nếu đưa vào khai thác; age chỉ thiếu 20% nên có thể điền bù.
