### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/titanic.csv")
```

### Các bước xử lý của câu 1 → 5 (chuẩn bị dữ liệu)

```python
df["relatives"] = df["sibsp"] + df["parch"]
df["not_alone"] = (df["relatives"] > 0).astype(int)  # 1: đi cùng người thân, 0: đi một mình
```

### Câu 6: Tiền xử lý các thuộc tính thiếu dữ liệu và xóa bỏ các thuộc tính không cần thiết

```python
# Tách ký tự đầu của cabin (boong tàu) trước khi xóa cabin, thiếu -> 'U' (Unknown)
deck_map = {"A": 1, "B": 2, "C": 3, "D": 4, "E": 5, "F": 6, "G": 7, "T": 8, "U": 0}
df["deck"] = df["cabin"].fillna("U").str[0].map(deck_map).fillna(0).astype(int)
```

```python
# Xem xét thuộc tính ticket
df["ticket"].describe()
```

```python
# Xóa các thuộc tính không cần thiết
df = df.drop(columns=["body", "cabin", "boat", "home.dest", "ticket"])
```

```python
# Điền giá trị thiếu cho age: sinh ngẫu nhiên trong [mean - std, mean + std]
mean = df["age"].mean()
std = df["age"].std()
is_null = df["age"].isnull().sum()
rng = np.random.default_rng(42)
rand_age = rng.integers(int(mean - std), int(mean + std), size=is_null)
df.loc[df["age"].isnull(), "age"] = rand_age
```

```python
# Xem xét thuộc tính embarked
df["embarked"].describe()
```

```python
common_value = "S"
df["embarked"] = df["embarked"].fillna(common_value)
# fare thiếu 1 giá trị -> điền trung vị
df["fare"] = df["fare"].fillna(df["fare"].median())
```

```python
print(df.isnull().sum())
df.head()
```

### Nhận xét
- Đã tách ký tự boong (deck) từ cabin rồi xóa body, cabin, boat, home.dest; ticket có 929 giá trị khác nhau nên cũng bị xóa.
- age được điền ngẫu nhiên trong [mean - std, mean + std], embarked điền bằng mode 'S', fare thiếu 1 giá trị điền bằng trung vị → không còn giá trị thiếu.
