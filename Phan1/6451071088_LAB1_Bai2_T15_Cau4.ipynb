### Import thư viện và đọc dữ liệu

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")
%matplotlib inline

df = pd.read_csv("../Dataset/titanic.csv")
```

### Câu 4: Khảo sát sự liên quan giữa Cổng lên tàu (embarked), Loại vé (pclass) với khả năng sống sót, phân biệt theo Giới tính (sex)

```python
FacetGrid = sns.FacetGrid(df, row="embarked", height=3.5, aspect=1.6)
FacetGrid.map(sns.pointplot, "pclass", "survived", "sex", palette=None, order=[1, 2, 3], hue_order=["female", "male"])
FacetGrid.add_legend()
plt.show()
```

```python
df.pivot_table(index="embarked", columns="sex", values="survived", aggfunc="mean").round(3)
```

```python
sns.barplot(x="pclass", y="survived", data=df)
plt.title("Tỷ lệ sống sót theo loại vé")
plt.show()
```

```python
grid = sns.FacetGrid(df, col="survived", row="pclass", height=2.4, aspect=1.6)
grid.map(plt.hist, "age", alpha=0.5, bins=20)
grid.add_legend()
plt.show()
```

### Nhận xét
- Cổng lên tàu có liên quan tới khả năng sống sót và phụ thuộc giới tính: nữ lên tàu ở cổng C sống sót cao nhất (~90%), cổng Q thấp nhất (~62%); nam lên ở cổng C cũng có tỷ lệ sống sót cao hơn (~31%) so với Q (~11%) và S (~17%).
- Loại vé ảnh hưởng rõ rệt: vé hạng 1 có tỷ lệ sống sót cao nhất, vé hạng 3 có tỷ lệ không sống sót lớn nhất (biểu đồ histogram tuổi theo pclass xác nhận điều này).
